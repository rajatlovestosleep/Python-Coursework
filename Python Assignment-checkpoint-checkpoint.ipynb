8: Creating New Columns

```python
#To create new columns by applying basic arithmetic operations on existing columns.
df['content_age'] = 2025 - df['release_year']
df[['title','release_year','content_age']][0:11]
```

9: Grouping and Aggregating

```python
#To group total titles by type
group_type = df.groupby('type')['title'].count()
print(group_type)
```

```python
# To get average content age by type
avg_age_by_type = df.groupby('type')['content_age'].median()
print(avg_age_by_type)
```

Explanation: groupby + aggregation compute counts and central tendency per group.

Inference: Quickly shows which type has more titles and relative age of content per type.

10: Joining DataFrames

```python
# To create small movies table and then inner join to original to demonstrate how joining of dataframes works
movies_small = df[df['type'] == 'Movie'][['show_id','title','release_year']].copy()
joined = pd.merge(movies_small, df, on='show_id', how='inner')
joined[0:10]
```

Explanation: pd.merge demonstrates joining a subset back to the main frame using show_id.
Inference: Join confirms relational operations (useful for combining preprocessed tables).

11: Re-Naming Columns

```python
#To rename multiple columns for clarity, using rename() method
df = df.rename(columns={
    'listed_in': 'genres',
    'release_year': 'year_released',
})
```

```python
# To verify the change
df.columns
```

Explanation: rename returns a new DataFrame mapping old names to new ones for readability.

Inference: Standardized column names improve readability and prevent confusion in grading.

12: Applying Operations to Multiple Columns

```python
#To strip whitespace and standardize case for several text columns
to_clean = ['type','country','rating','genres']
present = [c for c in to_clean if c in df.columns]
df[present] = df[present].apply(lambda col: col.astype('string').str.strip())
df[present][0:10]
```

Explanation: .apply() on a DataFrame applies function column-wise to multiple columns in one step.

Inference: Bulk-cleaning text columns removes formatting noise (extra spaces, inconsistent casing).

13: Filtering Records Based on Conditions

```python
# filter: TV shows added after 2019
tv_recent = df[(df['type'] == 'TV Show') & (df['year_released'] > 2019)]
tv_recent[['title','year_released']].head()
```

Explanation: Boolean masks combine conditions for targeted subsets.

Inference: Filtering isolates interesting slices (e.g., long movies or recently added TV shows).

14: Removing a Column or a Row from a Dataset

```python
# Creating a copy of the dataframe before removing columns
df_col_removed = df.copy()

# Removing the 'show_id' column as it is not relevant for analysis
df_col_removed.drop(columns=['show_id'], inplace=True)
df_col_removed.head()
```

```python
# Creating another copy before removing rows
df_row_removed = df.copy()

# Removing rows with missing 'title' entries
df_row_removed.dropna(subset=['title'], inplace=True)
df_row_removed.reset_index(drop=True, inplace=True)
df_row_removed.head()
```

Explanation:
Before performing any removal operation, copies of the original DataFrame were created.
This prevents permanent changes to the main dataset and avoids errors caused by inplace modifications.
Unnecessary columns like show_id were removed, and rows missing crucial information (title) were dropped from the copied DataFrames.

Inference:
Separate cleaned versions of the dataset were obtained —
one with redundant columns removed, and another with incomplete rows excluded.
This ensures safer data preprocessing without altering the main dataset.
