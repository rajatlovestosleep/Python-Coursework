# Importing Neccesary Libraries

```python
import numpy as np
```

## Creating a random array in numpy

```python
a = np.array(([202,535,960,370,206,171],
              [800,120,714,221,566,314],
              [430,558,187,472,199 ,971],
              [763,230,761,408,869,443],
              [591,513,905,485,291,376]))
```

## Finding out shape and size of a numpy array

```python
# Q1
print(a)
print(a.shape)
print(a.size)
print(a.dtype)
```

## Using Indexing to extract data from a numpy array

```python
# Q2
print(a[:1])
print(a[:,3:])
print(a[2,3])
```

## Performing mathematical operations on a numpy array

```python
#Q3
print(a*1.1)
a.sum(axis=0)
a.sum(axis=1)
```

## Applying statistical methods to a numpy array

```python
#Q4
print(a.mean())
print(np.median(a))
print(np.std(a))

print(np.argmax(a.mean(axis=1)))
print(np.argmin(a.sum(axis=0)))
```

## Extracting data from a numpy array by applying conditions

```python
#Q5
print(a[a>800])
np.where(np.any(a<200,axis=1))
```

## Reshaping a numpy array

```python
a.reshape(5,2,3)
```

## Creating a list for further use in stacking in a numpy array

```python
new_store=[123,456,789,101,112,678]
```

## Stacking the new list on our pre-existing array

```python
z=np.vstack((a,new_store))
print(z)
np.shape(z)
```

## Inserting a number in a random position in our original array

```python
print(np.insert(a,1,5))
```

## Extracting two rows from the original array

```python
h=a[0]
j=a[1]
print(h)
print(j)
```

## Taking dot product of the two extracted rows

```python
print(np.dot(h,j))
```

## Taking transpose of the original array

```python
print(a.T)
```

## Creating another random array and finding out its eigen-vectors

```python
d=np.array(([2,1],
            [0,1]))
print(d)
np.linalg.eig(d)
```
